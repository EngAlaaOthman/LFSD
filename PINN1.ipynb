```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Parameters
T_env = 25  # Ambient temperature in degrees Celsius
T0 = 90     # Initial temperature of the coffee
r_values = [0.05, 0.1, 0.2]  # Different cooling rates in min^-1

# Time values
t = np.linspace(0, 60, 100)  # Time from 0 to 60 minutes
```

```python
# Plotting
plt.figure(figsize=(10, 6))

# Loop through the different cooling rates
for r in r_values:
    # Temperature function for each cooling rate
    T = T_env + (T0 - T_env) * np.exp(-r * t)
    plt.plot(t, T, label=f'Cooling Rate r={r}', linestyle='-', linewidth=2)

# Add ambient temperature line
plt.axhline(y=T_env, color='r', linestyle='--', label='Ambient Temperature')

# Title and labels
plt.title('Cooling of a Coffee Cup Over Time for Different Cooling Rates')
plt.xlabel('Time (minutes)')
plt.ylabel('Temperature (°C)')
plt.legend()
plt.grid()
plt.xlim(0, 60)
plt.ylim(20, 95)
plt.show()
```
